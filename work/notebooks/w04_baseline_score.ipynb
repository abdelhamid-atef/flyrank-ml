## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Baseline score
df["baseline_score"] = (
    (df["days_since_last_update"] >= 180).astype(int) * 2 +
    (df["impressions_90d"] >= 500).astype(int) * 2 +
    (df["ctr"] < df["ctr"].median()).astype(int)
)

# Reason code
conditions = [
    (df["days_since_last_update"] >= 180),
    (df["ctr"] < df["ctr"].median()),
    (df["impressions_90d"] >= 500)
]

reasons = [
    "STALE_CONTENT",
    "LOW_CTR",
    "HIGH_IMPRESSIONS"
]

df["reason_code"] = "NORMAL"

for cond, reason in zip(conditions, reasons):
    df.loc[cond, "reason_code"] = reason

# Action
df["action"] = "Review Content"

# Sort
ranked = df.sort_values("baseline_score", ascending=False)

# Create folder
import os
os.makedirs("work/outputs", exist_ok=True)

# Save CSV
ranked.to_csv(
    "work/outputs/baseline_action_score.csv",
    index=False
)

print("CSV saved successfully.")

print(ranked[
    [
        "baseline_score",
        "reason_code",
        "action",
        "impressions_90d",
        "days_since_last_update",
        "ctr"
    ]
].head(20))
```
